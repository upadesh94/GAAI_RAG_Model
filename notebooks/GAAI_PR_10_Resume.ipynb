```python
from google.colab import userdata

# Make sure you have created a secret named 'GEMINI_API_KEY' in the Colab Secrets panel
GEMINI_API_KEY = userdata.get("GEMINI_API_KEY")

print("API Key loaded successfully!")
```

```python
from google import genai

client = genai.Client(api_key=GEMINI_API_KEY)

print("Gemini connected successfully!")
```

```python
from google.colab import files

uploaded = files.upload()

resume_path = list(uploaded.keys())[0]

print("Uploaded Resume:", resume_path)
```

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader(resume_path)

pages = loader.load()

print("Number of pages:", len(pages))
```

```python
print(pages[0].page_content)
```

```python
for i, page in enumerate(pages):
    print(f"\n--- Page {i+1} ---")
    print(page.page_content)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

chunks = text_splitter.split_documents(pages)

print("Number of chunks:", len(chunks))
```

```python
for i, chunk in enumerate(chunks):
    print(f"\n--- Chunk {i+1} ---")
    print(chunk.page_content)
```

```python
class GeminiEmbeddings:
    def __init__(self, client, model="gemini-embedding-2"):
        self.client = client
        self.model = model

    def embed_documents(self, texts):
        # LangChain expects this to return a list of lists of floats: List[List[float]]
        # If we pass a list of strings, we map them individually to avoid size/batch mismatches
        return [self.embed_query(text) for text in texts]

    def embed_query(self, text):
        result = self.client.models.embed_content(
            model=self.model,
            contents=text
        )
        # Extract the flat list of floats representing the embedding vector
        return result.embeddings[0].values
```

```python
embedding_model = GeminiEmbeddings(client)

print("Gemini Embedding Model Created!")
```

```python
test_embedding = embedding_model.embed_query("What are the skills in this resume?")

print("Embedding length:", len(test_embedding))
```

```python
print("ChromaDB is ready!")
```

```python
from langchain_community.vectorstores import Chroma

# Re-initialize the model to pick up the updated class definition
embedding_model = GeminiEmbeddings(client)

# Store in ChromaDB using from_documents
vector_db = Chroma.from_documents(
    documents=chunks,
    embedding=embedding_model,
    persist_directory="/content/resume_db"
)

print("Resume stored in ChromaDB successfully!")
```

```python
retriever = vector_db.as_retriever(
    search_kwargs={"k": 4}
)

print("Retriever created successfully!")
```

```python
question = "What programming languages does the candidate know?"

retrieved_chunks = retriever.invoke(question)

print("Number of retrieved chunks:", len(retrieved_chunks))
```

```python
for i, chunk in enumerate(retrieved_chunks):

    print(f"\n--- Retrieved Chunk {i+1} ---")
    print(chunk.page_content)
```

```python
MODEL_NAME = "gemini-3.8-flash"

print("LLM Model:", MODEL_NAME)
```

```python
def ask_gemini(prompt):

    response = client.models.generate_content(
        model=MODEL_NAME,
        contents=prompt
    )

    return response.text
```

```python
print(ask_gemini("What is Artificial Intelligence?"))
```

```python
def get_answer(question):

    # Retrieve relevant chunks
    retrieved_chunks = retriever.invoke(question)

    # Combine retrieved chunks
    context = ""

    for chunk in retrieved_chunks:
        context += chunk.page_content + "\n"

    # Prompt
    prompt = f"""
You are a Resume Question Answering Assistant.

Answer the question using ONLY the information
available in the provided resume context.

Do not make up any information.

If the answer is not available in the resume,
say:

"Information not available in the resume."

Resume Context:
{context}

Question:
{question}

Give a clear and concise answer.
"""

    # Gemini response
    response = client.models.generate_content(
        model=MODEL_NAME,
        contents=prompt
    )

    return response.text
```

```python
question1 = "What is the candidate's educational qualification?"

print("Question:", question1)
print("\nAnswer:")
print(get_answer(question1))
```

```python
question2 = "What technical skills does the candidate have?"

print("Question:", question2)
print("\nAnswer:")
print(get_answer(question2))
```

```python
question3 = "What projects has the candidate worked on?"

print("Question:", question3)
print("\nAnswer:")
print(get_answer(question3))
```
